```python
# Standard imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import warnings

warnings.filterwarnings('ignore')

# Plotting settings
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print('✓ Libraries loaded successfully')
```

## 1. Dataset Overview

```python
# Load raw hotel data
data_path = Path('../data/raw/hotels.csv')
df = pd.read_csv(data_path)

print(f'Dataset Shape: {df.shape}')
print(f'\nRows: {df.shape[0]:,} bookings')
print(f'Columns: {df.shape[1]} fields')
print(f'\nMemory Usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB')

# Data types
print('\n' + '='*50)
print('Data Types Summary')
print('='*50)
print(df.dtypes)
```

```python
# First few rows
print('First 5 rows:')
df.head()
```

## 2. Data Profiling

```python
# Comprehensive data info
print('Column Information:')
print('='*80)
for col in df.columns:
    non_null = df[col].notna().sum()
    null_pct = (df[col].isna().sum() / len(df)) * 100
    dtype = df[col].dtype
    
    print(f'{col:20} | {dtype:12} | Non-null: {non_null:7,} ({100-null_pct:5.1f}%) | Null: {null_pct:5.1f}%')
```

## 3. Missing Value Analysis

```python
# Missing values summary
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Column': missing.index,
    'Missing Count': missing.values,
    'Percentage': missing_pct.values
}).sort_values('Missing Count', ascending=False)

print(missing_df[missing_df['Missing Count'] > 0])
print(f'\nTotal missing values: {missing.sum():,}')
```

```python
# Visualize missing values
fig, ax = plt.subplots(figsize=(12, 6))
missing_pct_sorted = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
missing_pct_sorted[missing_pct_sorted > 0].plot(kind='barh', ax=ax, color='salmon')
ax.set_xlabel('Percentage Missing (%)')
ax.set_title('Missing Values by Column')
plt.tight_layout()
plt.show()

print(f'Columns with missing data: {(missing_pct > 0).sum()}')
```

## 4. Statistical Summaries

```python
# Numeric columns statistics
print('Numeric Columns Summary:')
print('='*80)
df.describe()
```

```python
# Categorical columns summary
print('\nCategorical Columns Summary:')
print('='*80)
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    print(f'\n{col}:')
    print(f'  Unique values: {df[col].nunique()}')
    print(f'  Top 5 values:')
    print(df[col].value_counts().head())
```

## 5. Data Visualization

```python
# Distribution of numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
n_cols = min(4, len(numeric_cols))

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols[:4]):
    axes[idx].hist(df[col].dropna(), bins=50, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'Distribution: {col}')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## 6. Data Quality Assessment

```python
# Data quality metrics
print('Data Quality Assessment:')
print('='*80)

# Duplicates
duplicates = df.duplicated().sum()
print(f'Total duplicate rows: {duplicates:,} ({(duplicates/len(df)*100):.2f}%)')

# Invalid values example
print(f'\nSample quality checks:')
print(f'  Rows checked: {len(df):,}')
print(f'  Complete rows (no nulls): {df.dropna().shape[0]:,}')
print(f'  Data completeness: {(df.dropna().shape[0]/len(df)*100):.1f}%')
```
